```python
import cv2
```

```python
from google.colab.patches import cv2_imshow

image = cv2.imread("/content/images.jpg")
cv2_imshow(image)
```

```python
print("shape:", image.shape)
#463 => height
#703 = > width
#3 => color channels(RGB)
```

```python
(b, g,r) = image[100, 50]
print(f"Blue: {b}, Green: {g}, Red: {r}")
```

```python
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
cv2_imshow(gray)
```

```python
resized = cv2.resize(gray, (300, 300))
cv2_imshow(resized)
```

```python
cropped = image[100:300, 150:350]
cv2_imshow(cropped)
```
