```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("StudentsPerformance.csv")

print(f"Dataset shape (rows, columns): {df.shape}")
print(f"Column name: {df.columns.tolist()}")
print(f"First 5 rows of the dataset:")
print(df.head())
```

```python
print("Averge score bu subject:")
df[["math score", "reading score", "writing score"]].mean()
```

```python
print("Maximum score bu subject:")
df[["math score", "reading score", "writing score"]].max()
```

```python
print("minimum score bu subject:")
df[["math score", "reading score", "writing score"]].min()
```

```python
print("Correlation:")
df[["math score", "reading score", "writing score"]].corr()
```
